```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('matches_1930_2022 (1).csv')

# --- Data Cleaning ---
# Fill missing numeric values for penalty shootouts with 0
df['home_penalty'] = df['home_penalty'].fillna(0)
df['away_penalty'] = df['away_penalty'].fillna(0)

# Convert these columns to integers
df['home_penalty'] = df['home_penalty'].astype(int)
df['away_penalty'] = df['away_penalty'].astype(int)

# For card columns, fill NaN with empty strings for consistency
card_columns = ['home_yellow_card_long', 'away_yellow_card_long', 'home_red_card', 'away_red_card', 'home_yellow_red_card', 'away_yellow_red_card']
for col in card_columns:
    df[col] = df[col].fillna('')

# --- Create New Columns ---

# 1. Total goals in the match (from open play)
df['total_goals'] = df['home_score'] + df['away_score']

# 2. Penalty shootout scores (already cleaned, just renaming for clarity)
df['home_penalty_score'] = df['home_penalty']
df['away_penalty_score'] = df['away_penalty']

# 3. Function to count entries in a cell split by the comma
def count_entries(cell):
    """
    Counts the number of entries in a cell.
    If the cell is a string with data, split by comma and count the elements.
    If the cell is empty or NaN, return 0.
    """
    if isinstance(cell, str) and cell != '':
        return len(cell.split(','))
    else:
        return 0

# 4. Calculate Yellow Cards
# Count the number of card entries in the 'long' description columns
df['home_yellow_cards'] = df['home_yellow_card_long'].apply(count_entries)
df['away_yellow_cards'] = df['away_yellow_card_long'].apply(count_entries)

# 5. Calculate Red Cards
# Sum direct red cards and red cards from two yellows
df['home_red_cards'] = df['home_red_card'].apply(count_entries) + df['home_yellow_red_card'].apply(count_entries)
df['away_red_cards'] = df['away_red_card'].apply(count_entries) + df['away_yellow_red_card'].apply(count_entries)

# --- Display the cleaned DataFrame with new columns ---
# Select a subset of columns to show the key original and new data
columns_to_display = [
    'home_team', 'away_team', 'home_score', 'away_score', 'total_goals',
    'home_penalty_score', 'away_penalty_score',
    'home_yellow_cards', 'away_yellow_cards',
    'home_red_cards', 'away_red_cards',
    'Round', 'Year'
]

print("Cleaned data with new columns (showing first 5 rows):")
print(df[columns_to_display].head())
df.to_csv('cleaned_world_cup_matches.csv', index=False)

# --- Optional: Save the cleaned DataFrame to a new CSV file ---
# df.to_csv('cleaned_world_cup_matches.csv', index=False)
# print("\nCleaned data saved to 'cleaned_world_cup_matches.csv'")
```
